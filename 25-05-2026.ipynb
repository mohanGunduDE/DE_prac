```python
from pyspark.sql import SparkSession

from pyspark.sql.functions import *
from pyspark.sql.window import Window

spark = SparkSession.builder.getOrCreate()

data = [
    (1, "Alice", "HR", 50000),
    (2, "Bob", "IT", 70000),
    (3, "Charlie", "IT", 80000),
    (4, "Diana", "Finance", 65000),
    (5, "Evan", "HR", 52000),
    (6, "Fay", "Finance", 72000),
    (7, "George", "IT", 90000)
]

columns = ["emp_id", "name", "department", "salary"]

df = spark.createDataFrame(data, columns)

df_cum= df.withColumn("cum_sum", sum("salary").over(Window.partitionBy("department").orderBy("salary").rowsBetween(Window.unboundedPreceding, Window.currentRow)))
display(df_cum)
```

```python
from pyspark.sql.functions import *

df2 = df.groupBy("department").agg(count("emp_id").alias("emp_count"), avg("salary").alias("avg_sal"))
df2.show()
```

```python

from pyspark.sql.functions import *
from pyspark.sql.window import Window
data = [
    (1, "Laptop", 2, 2000),
    (2, "Mobile", 5, 5000),
    (3, "Tablet", 3, 1500),
    (4, "Laptop", 1, 1000),
    (5, "Mobile", 2, 2000),
    (6, "Headphones", 10, 1000),
    (7, "Tablet", 4, 2000),
    (8, "Laptop", 3, 3000),
    (9, "Headphones", 5, 500),
    (10, "Monitor", 6, 1800)
]

columns = ["sale_id", "product", "quantity", "sales_amount"]

df_products = spark.createDataFrame(data, columns)

df_new = df_products.withColumn("new_sub",col("sales_amount")- lead("sales_amount")\
                                .over(Window.partitionBy("product").orderBy(col("sales_amount").desc())) )       

display(df_new)
```

```python

result_df = df_products.groupBy("product")\
        .agg(sum("sales_amount").alias("total_sales"))\
        .orderBy(col("total_sales").desc()).limit(3)
display(result_df)
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

data = [
    (1, "Alice", "2025-12-01"),
    (2, "Bob", "2026-03-10"),
    (3, "Charlie", "2024-11-15"),
    (4, "Diana", "2026-04-20"),
    (5, "Evan", "2025-08-01")
]

columns = ["emp_id", "name", "joining_date"]

df = spark.createDataFrame(data, columns)

from pyspark.sql.functions import to_date
df = df.withColumn("joining_date", to_date("joining_date")).withColumnRenamed("joining_date","JD")

df.filter(col("JD")>(date_sub(current_date(), 180))).show()
```

```python
customers = spark.createDataFrame(
[(1, "Alice"), (2, "Bob"), (3, "Charlie")], ["customer_id", "name"]
)
orders = spark.createDataFrame(
[(1, 100), (1, 200), (2, 150)], ["customer_id", "order_amount"]
)

df_join = customers.alias("a").join(orders.alias("b"), on = "customer_id", how = "right")
display(df_join)
```
