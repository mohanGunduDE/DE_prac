```python
dbutils.widgets.text("catalog_name","cab_cat")
cat_name = dbutils.widgets.get("catalog_name")
```

```python
spark.sql(f"create catalog {cat_name}")
```

```python
spark.sql(f"create schema if not exists {cat_name}.bronze")
spark.sql(f"create schema if not exists {cat_name}.silver")
spark.sql(f"create schema if not exists {cat_name}.gold")
```

```python
spark.sql(f"create volume if not exists {cat_name}.bronze.emp_vol")
```

```python
from pyspark.sql.types import *


schema = StructType([
    StructField("employee_id", IntegerType(), False),
    StructField("name",        StringType(),  False),
    StructField("department",  StringType(),  False),
    StructField("salary",      IntegerType(), False),
])



employees_data = [
    (1, "Alice",   "IT",      60000),
    (2, "Bob",     "HR",      52000),
    (3, "Charlie", "IT",      70000),
    (4, "Diana",   "Finance", 65000),
]



df = spark.createDataFrame(data =employees_data,schema=schema)
df.write.saveAsTable("cab_data_catalog.bronze.employees_bronze")
```

```python

from pyspark.sql.types import StructType, StructField, IntegerType, DateType
from pyspark.sql.functions import to_date

sales_schema = StructType([
    StructField("sale_id",     IntegerType(), False),
    StructField("employee_id", IntegerType(), False),
    StructField("amount",      IntegerType(), False),
    StructField("date",        StringType(),  False),  # parse to date below
])

sales_data = [
    (101, 1, 500, "2025-01-10"),
    (102, 2, 300, "2025-01-11"),
    (103, 1, 450, "2025-01-12"),
    (104, 3, 700, "2025-01-13"),
]

sales_df_raw = spark.createDataFrame(sales_data, schema=sales_schema)
sales_df = sales_df_raw.withColumn("date", to_date("date"))  # cast to DateType
sales_df.write.saveAsTable("cab_data_catalog.bronze.sales_bronze")
```
