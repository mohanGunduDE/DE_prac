```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window

Data = [
  ["ABC", "aa"],
  ["CDE", "cd"],
  ["EFG", "ef"],
  ["EFG", "aa"],
  ["ABC", "bb"],
  ["ABC", "cc"]
]

Schema = ['colA','colB']
```

```python
df = spark.createDataFrame(Data,Schema)
display(df)
```

```python

df2 = df.withColumn("rank", row_number().over(Window.partitionBy("colA").orderBy(col("colA"))))
display(df2)
```

```python
df3 = df2.groupBy("colA").pivot("rank").agg(first("colB"))
df3.select("colA", col("1").alias("F_COL"), col("2").alias("S_COL"), col("3").alias("3_COL")).show()
```
