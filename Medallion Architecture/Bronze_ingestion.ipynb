```python
from pyspark.sql.functions import *

df = spark.table("bronze_events")
df_silver = df.select(col("id").alias("user_id"), "ts","user.email","user.id","user.attrs.channel","user.attrs.region","campaign.discountPct","campaign.name","items")

df_silver = df_silver.withColumn("item_explode", explode_outer("items"))
display(df_silver)
df_silver.drop("items").write.mode("append").saveAsTable("cab_data_catalog.bronze.silver_events")

```
