```python
data = [
    (1, "Sales", "In-Progress", "25Dec2025", "02Jan2026"),
    (1, "Sales", "In-Progress", "10Jan2026", "15Jan2026"),
    (1, "Sales", "In-Progress", "20Jan2026", "25Jan2026")
]

columns = ["PId", "ProcessName", "P_Status", "P_startDate", "P_endDate"]

spark.createDataFrame(data= data, schema=columns).write.mode("overwrite").saveAsTable("cab_data_catalog.bronze.processes")
```
