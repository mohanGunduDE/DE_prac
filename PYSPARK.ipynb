```python
dep = "/Volumes/workspace/databricks_prac/prac_1/department.csv"
emp = "/Volumes/workspace/databricks_prac/prac_1/employee.csv"

df_dep = spark.read.format("csv") \
        .option("header", "true") \
        .option("inferSchema", "true") \
        .load(dep)

df_emp = spark.read.format("csv") \
        .option("header", "true") \
        .option("inferSchema", "true") \
        .load(emp)  

display(df_dep)
display(df_emp)
   
```

```python
df = df_dep.join(df_emp, df_dep.user == df_emp.emp_id, "inner")
display(df)
```
