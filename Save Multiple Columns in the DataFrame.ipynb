```python
 
data = [
    (1, "Sagar", 23, "Male", 68.0),
    (2, "Kim", 35, "Female", 90.2),
    (3, "Alex", 40, "Male", 79.1),
]
schema = "Id int,Name string,Age int,Gender string,Marks float"
df = spark.createDataFrame(data, schema)
```

```python
display(df)
```

```python
for i in df.dtypes:
    print(i)
```

```python
types_list = set([i[1] for i in df.dtypes])
types_list
```

```python
for i in types_list:
    col = []
    for j in df.dtypes:
        if i == j[1]:
            col.append(j[0])
    print(col)
    df.select(col).write.mode("append").saveAsTable(f"table_{i}")
            
```
