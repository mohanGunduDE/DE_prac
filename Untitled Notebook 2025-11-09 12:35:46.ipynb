```python
data = [('Manish', '{"street": "123 St", "city": "Delhi"}'),
        ('Ram', '{"street": "456 St", "city": "Mumbai"}')]
schema = ['name', 'address']

df_data = spark.createDataFrame(data,schema)
display(df_data)
```

```python
from pyspark.sql.functions import from_json, col

df_data = df_data.withColumn("new_Add",from_json(col("address"),"street string, city string"))
display(df_data)
```

```python
display(df_data.select("address","name","new_Add",col("new_Add").city.alias("city"),col("new_Add").street.alias("street")))
```

```python
data = [['2024-01-01', 20000],
        ['2024-01-02', 10000],
        ['2024-01-03', 150000],
        ['2024-01-04', 100000],
        ['2024-01-05', 210000]]

# define column names
columns = ['date', 'sales']

df_sal = spark.createDataFrame(data, columns)
display(df_sal)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window


df_sal = df_sal.withColumn("cum",sum("sales").over(Window.orderBy(desc("date")))) \
                .withColumn("lag",lag("sales").over(Window.orderBy(desc("date")))) \
                .withColumn("lead",lead("sales").over(Window.orderBy(desc("date"))))
display(df_sal)
```
