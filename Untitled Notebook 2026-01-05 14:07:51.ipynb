```python
Data = [
  ["ABC", "aa"],
  ["CDE", "cd"],
  ["EFG", "ef"],
  ["EFG", "aa"],
  ["ABC", "bb"],
  ["ABC", "cc"]
]

Schema = ['colA','colB']

df = spark.createDataFrame(Data,Schema)
display(df)
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *


df1 = df.withColumn("rn",row_number().over(Window.partitionBy(df.colA).orderBy(df.colA)))
display(df1)
df3 = df1.groupBy('colA').pivot('rn').agg(first('ColB'))
display(df3)
```
