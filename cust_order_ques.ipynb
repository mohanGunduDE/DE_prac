```python
customer_data = [(1, 'Manish'), (2, 'Rahul'), (3, 'Momu'), (4, 'Ram')]
schema = ['Customer_ID', 'Customer_Name']

df_cust = spark.createDataFrame(customer_data, schema)

display(df_cust)

```

```python
order_data = [(1, 4), (3, 2)]
schema = ['Order_ID', 'Customer_ID']

df_order = spark.createDataFrame(order_data, schema)

display(df_order)
```

```python
df_no_ord = df_cust.join(df_order, df_cust.Customer_ID == df_order.Customer_ID, 'left_anti')
df_order= df_cust.join(df_order, df_cust.Customer_ID == df_order.Customer_ID, 'left_semi')

display(df_no_ord)
display(df_order)
```
