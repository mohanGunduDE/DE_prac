```python
emp_data = [('Manish', 1, 75000),
            ('Raghav', 1, 85000),
            ('surya', 1, 80000),
            ('virat', 2, 70000),
            ('rohit', 2, 75000),
            ('jadeja', 3, 85000),
            ('anil', 3, 55000),
            ('sachin', 3, 55000),
            ('zahir', 4, 60000),
            ('bumrah', 4, 65000)]
schema = ['emp_name', 'dept_id', 'salary']

df_emp = spark.createDataFrame(emp_data,schema)
display(df_emp)
```

```python
dept_data = [(1, 'DATA ENGINEER'), (2, 'SALES'), (3, 'SOFTWARE'), (4, 'HR')]
schema1 = ['dept_id', 'dept_name']

df_dpt = spark.createDataFrame(dept_data,schema1)
df_dpt = df_dpt.withColumnRenamed("dept_id","dept_id1")
```

```python
display(df_emp)
display(df_dpt)
```

```python
df_join = df_emp.join(df_dpt,df_emp.dept_id == df_dpt.dept_id1,"left")
display(df_join.groupBy("dept_name").agg(max("salary")))
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window

df_join = df_join.withColumn("rn", row_number().over(Window.partitionBy("dept_id").orderBy(desc("salary")))).filter(col("rn") == 1)

display(df_join)
```
