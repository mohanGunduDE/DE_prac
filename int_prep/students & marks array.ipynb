```python
from pyspark.sql.types import *

data = [
    (1, "Alice", ["Math", "Science", "English"], [85, 90, 78]),
    (2, "Bob", ["Math", "Science", "English"], [88, 76, 92]),
    (3, "Charlie", ["Math", "Science", "English"], [65, 70, 80])
]


schema = StructType([
    StructField("id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("subjects", ArrayType(StringType()), True),
    StructField("subject_marks", ArrayType(IntegerType()), True)
])


df = spark.createDataFrame(data, schema)
```

```python
display(df)
```

```python
from pyspark.sql.functions import *

df_test = df.withColumn("sub_test", explode("subjects")).withColumn("mrk_test", explode("subject_marks"))
display(df_test)
```

```python
from pyspark.sql.functions import *
df1 = df.withColumn("combined", arrays_zip("subjects", "subject_marks"))
display(df1)
df1 = df1.withColumn("explodede", explode("combined"))

df1.select("id","name","combined",col("explodede.subjects").alias("subjects"),col("explodede.subject_marks").alias("subject_marks")).show(truncate=False)
```
