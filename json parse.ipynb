```python
path = "/Volumes/transportation/bronze/source_data/customers_nested.json"

df= spark.read.format("json").option("multiline", True).load(path)
display(df.printSchema())
```

```python
display(df)
```

```python
df_1= df.select("customer_id","name","orders",explode("orders").alias("orders_exp"))
df_2 = df_1.select("customer_id","name","orders_exp.order_id",explode("orders_exp.items").alias("items"))
df_3 = df_2.select("customer_id","name","order_id","items.item_id","items.product")
display(df_3)
```

```python
from pyspark.sql.functions import *

df_new = df.withColumn("orders_explode", explode("orders"))
df_new2 = df_new.withColumn("order_id", col("orders_explode.order_id"))\
                .withColumn("items", explode(col("orders_explode.items")))\
                .withColumn("item_id",col("items.item_id"))\
                .withColumn("product",col("items.product"))
df_new2.select("orders","order_id","items","item_id","product").show(truncate=False)
```

```python
from pyspark.sql.functions import *

df_exp = df.withColumn("order_id", explode("orders").getItems(1))
display(df_exp)
```

```python
df_initial = df.select("customer_id","name",explode("orders").alias("orders"))
df_second = df_initial.select("customer_id","name",col("orders.order_id").alias("order_id"),explode("orders.items").alias("items"))
df_final = df_second.select("customer_id","name","order_id",col("items.item_id").alias("item_id"),col("items.product").alias("product"))
display(df_final)
```
