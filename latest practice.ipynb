```python
def prime(n):
    if n<=1:
        return False
    else :
        for i in range(2,n):
            if n%i == 0:
                return False
            else:
                return True

prime(13)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import StructType,StructField,StringType,IntegerType

source = "/Volumes/cab_data_catalog/bronze/corrupt/prac.csv"

schema = StructType([
    StructField("First_name",StringType()),
    StructField("Last_name",StringType()),
    StructField("Pin",IntegerType()),
    StructField("Address",StringType())
])

df = spark.read.format("csv").schema(schema).option("header",True).load(source)
df1 = df.withColumn("name", concat(col("First_name"),lit("_"),col("Last_name")))\
        .withColumn("address_new",when(col("Pin").isin(1111),col("Address")).otherwise(None))\
        .withColumn("Address",when(col("Address").isin("Unknown"),None).otherwise(col("Address")))

display(df1)
```

```python
from pyspark.sql import SparkSession
from pyspark.sql.window import Window
import pyspark.sql.functions as F

data = [
    (1, "Bitcoin", "2024-02-01", 430, 435),
    (1, "Bitcoin", "2024-02-02", 435, 440),
    (1, "Bitcoin", "2024-02-03", 440, 439),
    (1, "Bitcoin", "2024-02-04", 445, 430),
    (2, "Ethereum", "2024-02-01", 320, 325),
    (2, "Ethereum", "2024-02-02", 325, 330),
    (2, "Ethereum", "2024-02-03", 330, 335),
    (2, "Ethereum", "2024-02-04", 335, 340),
    (3, "Solana", "2024-02-01", 120, 125),
    (3, "Solana", "2024-02-02", 125, 130),
    (3, "Solana", "2024-02-03", 130, 135),
    (3, "Solana", "2024-02-04", 135, 140),
]

columns = ["Coin_ID", "Coin_Name", "Date", "Open_Price", "Close_Price"]

df = spark.createDataFrame(data =data, schema = columns)
```

```python
df.withColumn("Date", F.to_date(F.col("Date"),"yyyy-mm-dd"))

df.show()
```

```python
df = df.withColumn("lag1",
              F.lag(F.col("Close_Price"))\
              .over(Window.partitionBy(F.col("Coin_ID")).orderBy(F.col("Date"))))\
   .withColumn("lag2",
              F.lag(F.col("Close_Price"),2)\
              .over(Window.partitionBy(F.col("Coin_ID")).orderBy(F.col("Date"))))

display(df)

```

```python
df_result = df.filter(
    (F.col("Close_Price") > F.col("lag1")) &
    (F.col("lag1") > F.col("lag2"))).select("Coin_ID", "Coin_Name").distinct()

df_result.display()
```

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window

# 1) Parse the US-style date safely.
#    Prefer try_to_date if available; otherwise use to_timestamp + cast with a guard.
# Option A: If your runtime supports try_to_date
df = df.withColumn("login_date", F.expr("try_to_date(login_date, 'M/d/yyyy')"))

# Option B: Portable approach if try_to_date isn't available
# df = df.withColumn(
#     "login_date",
#     F.to_timestamp("login_date", "M/d/yyyy").cast("date")
# )

# (Optional) If there are still nulls after parsing and you want to detect them:
# df.filter(F.col("login_date").isNull()).show(truncate=False)

# 2) Window for rank by employee ordered by parsed date
w = Window.partitionBy(F.col("emp_id")).orderBy(F.col("login_date"))

df = df.withColumn("rank", F.row_number().over(w))

# 3) Row-wise date subtraction using expr (rank is a column)
df = df.withColumn("updated_col", F.expr("date_sub(login_date, rank)"))

# 4) Show results in Databricks
# display(df)
df.show(truncate=False)
```
