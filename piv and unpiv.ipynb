```python
#unpiv = "/Volumes/workspace/databricks_prac/prac_1/unpiot.csv"
piv = "/Volumes/workspace/databricks_prac/prac_1/pivot_emp_dept.csv"

df_piv = spark.read.format("csv").option("Header","true").option("inferSchema", "true").load(piv)

df_piv_n = df_piv.select("emp_id","dept_name","dept_id","name","country")
df_piv_n.printSchema()
```

```python
#display(df_piv_n)

df_piv_n.groupBy("name").pivot("country","dept_name").sum("dept_id").show()
```
