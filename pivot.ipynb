```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window

Data = [
  ["ABC", "aa"],
  ["CDE", "cd"],
  ["EFG", "ef"],
  ["EFG", "aa"],
  ["ABC", "bb"],
  ["ABC", "cc"]
]

Schema = ['colA','colB']

df = spark.createDataFrame(Data,Schema)
display(df)
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window


df2 = df.withColumn("rn", row_number().over(Window.partitionBy('colA').orderBy('colA')))
#df2.groupBy("colA").pivot("colB").agg(sum("rn")).show()
df3 = df2.groupBy("colA").pivot("rn").agg(first("colB"))
df3.select("colA",col("1").alias("First_col"),col("2").alias("second_col"),col("3").alias("third_col")).show()

```

```python
df1 = df.withColumn("row_num", row_number().over(Window.partitionBy("colA").orderBy("colA")))
df2= df1.groupBy("colA").pivot("row_num").agg(first("colB"))
df3 = df2.select("colA",col("1").alias("First_col"),col("2").alias("second_col"),col("3").alias("third_col"))
display(df3)
```
