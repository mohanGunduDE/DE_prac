```python
Val = [(1,"aaa","ind"),(2,"bbb","aus"),(3,"ccc","ind"),(4,"ddd","aus")]
tab = ['id', 'name', 'country']
df = spark.createDataFrame(Val,tab)
df.show()
df.printSchema()
```

```python
from pyspark.sql.types import StructType,StructField,IntegerType,StringType

Val = [(1,"aaa","ind"),(2,"bbb","aus"),(3,"ccc","ind"),(4,"ddd","aus")]
tab = StructType([StructField(name = "id",dataType = IntegerType()),
                  StructField(name = "name",dataType = StringType()),
                  StructField(name = "country",dataType = StringType())])
df = spark.createDataFrame(Val,tab)
df.show()
df.printSchema()
```
