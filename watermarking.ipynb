```python
from pyspark.sql.functions import *
from pyspark.sql import Row
import datetime

# Current time
now = datetime.datetime.now()

data = [
    Row(id=1, value=100, event_time=now - datetime.timedelta(minutes=1)),   # normal
    Row(id=2, value=200, event_time=now - datetime.timedelta(minutes=3)),   # normal
    Row(id=3, value=300, event_time=now - datetime.timedelta(minutes=7)),   # late (should be dropped)
]

df = spark.createDataFrame(data)

```

```python
df.withWatermark("event_time","5 minutes").groupBy(window("event_time","5 minutes")).agg(count("*").alias("total"))
```

```python
df.writeStream.format("delta").option("checkpointLocation", "/Volumes/transportation/bronze/source_data/_checkpoints/wm_test_table").outputMode("append").toTable("transportation.bronze.wm_test_table")
```
