```python

data = [
    (1, 'mansi', 'india', "2022-01-01"),
    (2, 'ram', 'india', "2023-01-01"),
    (3, 'sunny', 'UK', "2024-01-01"),
    (4, 'sohan', 'UK', "2025-01-01")
]
columns = ['id', 'name', 'country', 'Date']

df = spark.createDataFrame(data, columns)
display(df)
```

```python

from pyspark.sql.functions import *
df.withColumn("Date_n", dayofyear("Date")).show()
```

```python

from pyspark.sql.functions import *
df = df.withColumn("New_date",date_add("Date",7)) \
    .withColumn("date_sub",date_sub("Date",14))
display(df)
```

```python
from pyspark.sql.functions import *

df = spark.createDataFrame(
    [("2025-11-06", "1993-01-04")],
    ["date1", "date2"]
)

df = df.withColumn(
    "diff_days",
    date_diff("date1", "date2")
)

display(df)
```
